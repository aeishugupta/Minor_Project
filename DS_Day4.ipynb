```python
#Data Visualization
# Importing matplotlib


# Introduction to Matplotlib
# Import Library
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# First plot
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

```python
# Add Title
plt.plot(x,y)
plt.title('Linear Graph')
plt.show()
```

```python
# Axis Labels
plt.plot(x,y)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.show()
```
